# loops(i) while

```python
print('data science')   #print 5 times datascience
print('data science')
print('data science')
print('data science')
print('data science')
```

```python
i=1     #initializing
while i<=5:  #condition    
    print('data science')
    i=i+1     #increment
```

```python
i=1
while i<=100:
    print('data science')
    i=i+1
```

```python
i=5
while i>=1:   #condition
    print('data science')    #same statements repeats is called iteration
    i=i-1  #decrement
```

```python
i=5
while i>=1:   #condition
    print('data science',i)    #same statements repeats is called iteration
    i=i-1  #decrement
```

```python
i=1     #initializing
while i<=5:      
    print('data science',i)
    i=i+1     #increment
```

```python
i=1     #initializing
while i<=5:      
    print('data science')
    j=1
    while j<=4:
        print('technology')
        j=j+1
    i=i+1     #increment
    print()
```

```python
i=1     #initializing
while i<=5:      
    print('data science',end='-')

    j=1
    while j<=4:
        print('technology',end='-')
        j=j+1
    i=i+1     #increment
    print()
```

# ii)for loop

```python
name="nit"
for i in name:     #strings with for loop
    print(i)
```

```python
l=[1,2,3,'nit']       #list with for loop
for i in l:
    print(i)
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,10,3):
    print(i)
```

```python
#print the value which is divisible by 5
for i in range(1,51):
    if i%5==0:
        print(i)
```

```python
for i in range(1,51):  #it skips the number which is divisible by 5
    if i%5!=0:
        print(i)
```

# i)break keyword

```python
for i in range(1,11):
    print(i)
```

```python
for i in range(1,11):
    if i==5:
        break        #it breaks the loop
    print(i)
```

```python
for i in range(1,11):
    if i==4:
        break        #it breaks the loop
    print("hello",i)
```

# ii)continue

```python
for i in range(1,11):
    if i==4:      #condition
        continue       #it continues the loop but skips the given condiiton
    print("hello",i)
```

```python
for i in range(1,11):
    pass                           #it skips the error,it s used i abstract class and abstract method in oops concept
```

```python
#write a program to print numbers from 1 to 50 ,but dont print the number which is divisible by 3 and 5

for i in range(1,51):
    if i%3==0 or i%5==0:           #it continues the loop but skips the number which is divisible by 3 and 5
        continue
    print(i)
print("end")
```

# for else

```python
nums=[12,15,18,21,26]
for num in nums:
    if num%5==0:
        print(num)
```

```python
nums=[12,15,18,21,26,20,30]
for num in nums:
    if num%5==0:
        print(num)
```

```python
nums=[12,14,18,21,20,25]
for num in nums:
    if num%5==0:
        print(num)
```

```python
nums=[12,14,18,21,20,25]
for num in nums:
    if num%5==0:
        print(num)    #it doesnt print another value which is divisible by 5 (breaks the loop)
        break
```

```python
nums=[7,14,18,21,23,27]
for num in nums:
    if num%5==0:
        print(num)    #no error no output becz there is no num available in list that is divisible by 5
        break
```

```python
nums=[12,14,18,21]
for num in nums:
    if num%5==0:
        print(num)    #here it is checking all the values (not a good method)
        break
    else:
        print("Number not found")
```

```python
nums=[12,14,18,21]
for num in nums:
    if num%5==0:
        print(num)    #here we are using for else
        break
else:
    print("Number not found")
```

```python
nums=[12,14,18,21,30]
for num in nums:
    if num%5==0:
        print(num)    #here we are using for else
        break
else:
    print("Number not found")
```

```python
# write a program to find out whether the number is prime or not with forelse
num=13
for i in range(2,num):
    if num%i==0:
        print("Number is not a prime")
        break
else:
    print("Number is a prime")
```
