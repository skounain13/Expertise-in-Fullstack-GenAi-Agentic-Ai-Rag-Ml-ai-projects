# i)swapping using 3rd variables

```python
a1,b1=7,8
print(a1)
print(b1)
```

```python
temp=a1
a1=b1
b1=temp
print(a1)
print(b1)
```

# ii)Without using 3rd variable:

```python
a1,b1=7,8
a1,b1=b1,a1
print(a1)
print(b1)
```

```python
a2=5
b2=6
a2=a2+b2 #5+6=11
b2=a2-b2 #11-6=5
a2=a2-b2 #11-5=6
print(a2)
print(b2)
```
