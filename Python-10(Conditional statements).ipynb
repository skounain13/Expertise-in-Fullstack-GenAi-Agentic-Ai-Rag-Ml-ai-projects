# ii)if else

```python
x=8
r=x%2
if r==0:
    print('Even number') # we use else to save memory(only one condition(binary condition) if else should use)
else:
    print('Odd number')
```

# iii)Nested if

```python
x=3
r=x%2
if r==0:
    print('Even number') 
    if x>5:                             # here there are 2 condition we use nested if
        print('Greater number')
else:
    print('odd number')
```

```python
x=4
r=x%2
if r==0:
    print('Even number') 
    if x>5:                             
        print('Greater number')
else:
    print('odd number')
```

# nested if else

```python
x=4
r=x%2
if r==0:
    print('Even number') 
    if x>5:                             # here there are 2 condition we use nested if else (1st condition(even or odd) 2nd(greater or less)
        print('Greater number')            # if else inside if else is called as nested if else
    else:
        print('Less number')
else:
    print('odd number')
```
