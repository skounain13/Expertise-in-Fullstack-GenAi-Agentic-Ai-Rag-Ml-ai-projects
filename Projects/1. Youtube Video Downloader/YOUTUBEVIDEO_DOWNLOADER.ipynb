```python
from pytubefix import YouTube

link = "https://www.youtube.com/watch?v=2YV6-csi2zo"

print("Connecting to YouTube...")

yt = YouTube(link)

print(f"Title: {yt.title}")

print("Checking available streams...")

stream = (
    yt.streams
    .filter(progressive=True, file_extension="mp4")
    .order_by("resolution")
    .desc()
    .first()
)

if stream:
    print("Downloading video...")
    stream.download()
    print("Download completed successfully!")
else:
    print("No progressive MP4 stream available.")
```

```python
from pytubefix import YouTube

link = "https://www.youtube.com/watch?v=2YV6-csi2zo"

print("Connecting to YouTube...")

yt = YouTube(link)

print(f"Title: {yt.title}")

print("Selecting video and audio...")

video = (
    yt.streams
    .filter(adaptive=True, only_video=True, file_extension="mp4")
    .order_by("resolution")
    .desc()
    .first()
)

audio = (
    yt.streams
    .filter(only_audio=True, file_extension="mp4")
    .first()
)

if video and audio:
    print("Downloading video...")
    video.download(filename="video.mp4")

    print("Downloading audio...")
    audio.download(filename="audio.mp4")

    print("Merging video and audio...")

    import subprocess

    subprocess.run([
        "ffmpeg", "-y",
        "-i", "video.mp4",
        "-i", "audio.mp4",
        "-c:v", "copy",
        "-c:a", "aac",
        "final_video.mp4"
    ], check=True)

    print("Download completed successfully!")

else:
    print("Required video or audio stream not found.")
```
