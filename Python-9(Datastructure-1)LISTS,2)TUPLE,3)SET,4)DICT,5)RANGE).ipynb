# ALL/ANY

The all() method returns():
*True-if all keys of dictionary are true
*False -if any keys of the dictionary is false

The any() funciton returns True if any of the dictionary is True .If not, any() returns False.

```python
mydict1={'Name':'Asif','ID':12345,'DOB':1991,'Job':'Analyst'}
mydict1
```

```python
all(mydict1)  #will return false as one value is false(value 0)
```
