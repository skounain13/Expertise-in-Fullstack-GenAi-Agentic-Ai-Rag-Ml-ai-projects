```python
#What will be the output? 
 
def calculate(n): 
    if n <= 1: 
        return 1 
    return n + calculate(n - 2) 
 
print(calculate(5))
```

```python
#What will be the output? 
 
num = 121 
reverse = 0 
temp = num 
while temp > 0: 
    digit = temp % 10 
    reverse = reverse * 10 + digit 
    temp //= 10 
if num == reverse: 
    result = 'Palindrome' 
else: 
    result = 'Not Palindrome' 
print(result)
```

```python
#What will be the output? 
 
num = 29 
is_prime = True 
if num < 2: 
    is_prime = False 
else: 
    for i in range(2, int(num ** 0.5) + 1): 
        if num % i == 0: 
            is_prime = False 
            break 
print('Prime' if is_prime else 'Not Prime')
```
