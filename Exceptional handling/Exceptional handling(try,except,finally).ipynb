```python
# two print two statements:
c=3
d=0
try:
    print(c/d)
except Exception:
    print("Hey you can't divide number by zero")
print("bye")
```

```python
# to recognize the error
c=3
d=0
try:
    print(c/d)
except Exception as e:
    print("Hey you can't divide number by zero",e)
print("bye")
```

```python
# Opening and closing db:
c=3
d=0
try:
    print("db open")
    print(c/d)
    print("db closed")
except Exception as e:
    print("Hey you can't divide number by zero",':',e)
```

```python
# finally keyword:
c=3
d=3
try:
    print("db open")
    print(c/d)
except Exception as e:
    print("Hey you can't divide number by zero",':',e)
    print("db closed")
finally:
    print("resource closed")
```
