# iii)GLOBAL KEYWORD

```python
a=10
def something():
    global a                                #if we remove this variable then can default it consider as global variable
    print('in function',a)
something()
print('out function',a)
```

```python
x=10                      # global variable
def update_x():
    global x          #declare that we are using the global variable x
    x+=5                #modify the global variable
update_x()
print(x)   
```
